```python
import pandas as pd
import numpy as np
from yahpo_gym import BenchmarkSet
from package_hpi_parego.hpi_utils import sum_hps_concise
```

```python
# Read result file from Pyexperimenter database\n",
df = pd.read_csv('../../../database.csv')
df['hpi_res'] = df['hpi_res'].apply(eval)

df['imp_hps'] = df['hpi_res'].apply(sum_hps_concise, 0.8)
df['imp_hps_str'] = df['imp_hps'].astype(str)
df['weight'] = df['weight_vector'].apply(lambda x: round(eval(x)[0],1))
```

# LCBench

```python
group = df[df['scenario']=='lcbench'].groupby(['dataset','weight'])['imp_hps_str'].unique().reset_index().explode('imp_hps_str').groupby('dataset')['imp_hps_str'].value_counts().reset_index().groupby('dataset')
```

```python
# ids for evaluation
ids = set((group['count'].max()/df[df['scenario']=='lcbench'].groupby('dataset')['weight'].nunique()).sort_values().head(15).index).intersection(set(group.count().sort_values('imp_hps_str', ascending=False)['count'].head(12).index))
yahpo = BenchmarkSet(scenario="lcbench")
ids_eval = [yahpo.instances[id] for id in ids]
ids_eval
```

```python
# ids for ablation
ids = set((group['count'].max()/df[df['scenario']=='lcbench'].groupby('dataset')['weight'].nunique()).sort_values().head(16).index).intersection(set(group.count().sort_values('imp_hps_str', ascending=False)['count'].head(16).index))
yahpo = BenchmarkSet(scenario="lcbench")
ids_abl = [yahpo.instances[id] for id in ids]
print(set(ids_abl).difference(set(ids_eval)))
```
